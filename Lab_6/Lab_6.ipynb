```python
import heapq
def print_edges(edges):
    formatted_edges = [f"{u}-{v}" for u, v in edges]
    print("{", ", ".join(formatted_edges), "}")
print_edges([(1, 2), (2, 3), (3, 4)])
```

```python
def prufer_encode(n, edges, show_steps=True):
    
    adjacency = {
        node: set()
        for node in range(1, n + 1)
    }

    for u, v in edges:
        adjacency[u].add(v)
        adjacency[v].add(u)

    leaves = []

    for node in range(1, n + 1):
        if len(adjacency[node]) == 1:
            heapq.heappush(leaves, node)

    prufer_sequence = []

    for step in range(1, n - 1):


        leaf = heapq.heappop(leaves)


        neighbour = next(iter(adjacency[leaf]))


        prufer_sequence.append(neighbour)

        if show_steps:
            print(
                f"Step {step}: "
                f"remove leaf {leaf}, "
                f"record neighbour {neighbour}"
            )

        adjacency[leaf].remove(neighbour)
        adjacency[neighbour].remove(leaf)

        if len(adjacency[neighbour]) == 1:
            heapq.heappush(leaves, neighbour)

    return prufer_sequence
```

```python
test_edges = [
    (1, 2),
    (2, 3),
    (3, 4)
]

test_code = prufer_encode(
    n=4,
    edges=test_edges
)

print("Prüfer sequence:", test_code)
```

```python
def prufer_decode(sequence, show_steps=True):
    
    n = len(sequence) + 2

    
    degree = {
        node: 1
        for node in range(1, n + 1)
    }

    
    for node in sequence:
        degree[node] += 1

   
    leaves = []

    for node in range(1, n + 1):
        if degree[node] == 1:
            heapq.heappush(leaves, node)

    decoded_edges = []

    for step, current_node in enumerate(sequence, start=1):

        
        leaf = heapq.heappop(leaves)

        
        decoded_edges.append((leaf, current_node))

        if show_steps:
            print(
                f"Step {step}: "
                f"select leaf {leaf}, "
                f"create edge {leaf}-{current_node}"
            )

        
        degree[leaf] -= 1
        degree[current_node] -= 1

        
        if degree[current_node] == 1:
            heapq.heappush(leaves, current_node)

    
    first_node = heapq.heappop(leaves)
    second_node = heapq.heappop(leaves)

    decoded_edges.append((first_node, second_node))

    if show_steps:
        print(
            f"Final step: create edge "
            f"{first_node}-{second_node}"
        )

    return decoded_edges
```

```python
test_sequence = [2, 3]

decoded_tree = prufer_decode(test_sequence)

print("\nDecoded edges:")
print_edges(decoded_tree)
```

```python
def normalize_edges(edges):
    normalized = set()

    for u, v in edges:
        normalized.add(tuple(sorted((u, v))))

    return normalized
```

```python
edges_1 = [(1, 2), (2, 3)]
edges_2 = [(2, 1), (3, 2)]

print(normalize_edges(edges_1))
print(normalize_edges(edges_2))
```

```python
def verify_tree(original_edges, decoded_edges):


    original_set = normalize_edges(original_edges)
    decoded_set = normalize_edges(decoded_edges)

    return original_set == decoded_set
```

```python
original = [(1, 2), (2, 3), (3, 4)]
decoded = [(2, 1), (3, 2), (4, 3)]

result = verify_tree(original, decoded)

print("PASS" if result else "FAIL")
```

```python
def test_tree(tree_name, n, edges):
   

    print("=" * 60)
    print(tree_name)
    print("=" * 60)

    print("\nOriginal edges:")
    print_edges(edges)

    print("\nEncoding steps:")
    sequence = prufer_encode(
        n=n,
        edges=edges,
        show_steps=True
    )

    print("\nPrüfer sequence:")
    print(sequence)

    print("\nDecoding steps:")
    decoded_edges = prufer_decode(
        sequence=sequence,
        show_steps=True
    )

    print("\nDecoded edges:")
    print_edges(decoded_edges)

    is_correct = verify_tree(
        original_edges=edges,
        decoded_edges=decoded_edges
    )

    print("\nVerification:")
    print("PASS" if is_correct else "FAIL")

    return sequence, decoded_edges
```
