```python
import pandas as pd

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import OneHotEncoder
```

```python
encodings = ["utf-8", "utf-8-sig", "latin-1", "cp1252"]

for enc in encodings:
    try:
        df = pd.read_csv("spam.csv", encoding=enc)
        print(enc, "works")
    except:
        print(enc, "does not work")
```

```python
df = pd.read_csv("./spam.csv",encoding='latin-1')

df.head()
```

```python
df.shape
df.columns
```

```python
df = df[["v1", "v2"]]

df.head()
```

```python
df.columns = ["label", "message"]

df.head()
```

```python
df.isnull().sum()
```

```python
df["label"].value_counts()
```

```python
bow_vectorizer = CountVectorizer()

bow_matrix = bow_vectorizer.fit_transform(df["message"])
```

```python
bow_matrix.toarray()
```

```python
bow_words = bow_vectorizer.get_feature_names_out()

print(bow_words[:50])
```

```python
first_bow = bow_matrix[0].toarray()[0]

word_positions = first_bow.nonzero()[0]

bow_result = pd.DataFrame({
    "Word": bow_words[word_positions],
    "Count": first_bow[word_positions]
})

bow_result
```

```python
tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(df["message"])
tfidf_matrix.toarray()
```

```python
tfidf_words = tfidf_vectorizer.get_feature_names_out()

print(tfidf_words[:50])
```

```python
first_tfidf = tfidf_matrix[0].toarray()[0]

word_positions = first_tfidf.nonzero()[0]

tfidf_result = pd.DataFrame({
    "Word": tfidf_words[word_positions],
    "TF-IDF": first_tfidf[word_positions]
})

tfidf_result
```

```python
onehot_encoder = OneHotEncoder(sparse_output=False)

label_onehot = onehot_encoder.fit_transform(df[["label"]])
```

```python
onehot_df = pd.DataFrame(
    label_onehot,
    columns=onehot_encoder.get_feature_names_out(["label"])
)

onehot_df.head()
```
