```python
def welsh_powell(graph):
    degree = {}

    for vertex in graph:
        degree[vertex] = len(graph[vertex])

    order = sorted(graph, key=lambda x: degree[x], reverse=True)

    color = {}

    for vertex in order:
        used = []

        for neighbor in graph[vertex]:
            if neighbor in color:
                used.append(color[neighbor])

        c = 1
        while c in used:
            c = c + 1

        color[vertex] = c

    return degree, order, color


def show_result(degree, order, color):
    print("Degrees:")
    for v in degree:
        print(v, "=", degree[v])

    print("\nWelsh-Powell Order:")
    print(order)

    print("\nAssigned Colors / Slots:")
    for v in order:
        print(v, "->", color[v])

    print("\nTotal colors used:", max(color.values()))
```

```python
graph1 = {
    "A": ["B", "C", "D", "E"],
    "B": ["A", "C", "F"],
    "C": ["A", "B", "D", "H"],
    "D": ["A", "C", "E"],
    "E": ["A", "D", "F"],
    "F": ["B", "E", "G"],
    "G": ["F", "H"],
    "H": ["C", "G"]
}

degree, order, color = welsh_powell(graph1)
show_result(degree, order, color)
```

```python
graph2 = {
    "Math": ["Physics", "Chemistry"],
    "Physics": ["Math", "Chemistry", "Biology"],
    "Chemistry": ["Math", "Physics", "Biology"],
    "Biology": ["Physics", "Chemistry", "History"],
    "History": ["Biology", "Art", "Music"],
    "Art": ["History", "Music"],
    "Music": ["History", "Art"]
}

degree, order, color = welsh_powell(graph2)
show_result(degree, order, color)
```
